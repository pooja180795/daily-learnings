```python
import numpy as np
```

```python
train_data = np.loadtxt("pen+based+recognition+of+handwritten+digits/pendigits.tra", delimiter=",")
print(train_data.shape)
```

```python
test_data = np.loadtxt("pen+based+recognition+of+handwritten+digits/pendigits.tes", delimiter=",")
print(test_data.shape)
```

```python
X_train = train_data[:,0:16]
y_train = train_data[:, 16]
```

```python
print(X_train.shape)
print(y_train.shape)
```

```python
X_test = test_data[:,0:16]
y_test = test_data[:,16]
```

```python
print(X_test.shape)
print(y_test.shape)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier(n_neighbors = 3)
```

```python
knn.fit(X_train, y_train)
```

```python
y_pred = knn.predict(X_test)
```

```python
print(y_pred[:10],y_test[:10])
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test, y_pred)
```
